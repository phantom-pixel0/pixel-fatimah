```python
import python as py
num = int(input("enter a number: "))

if num % 2 == 1:
    print(num, "is odd")
else:
    print(num, "is even")
```

```python
a = float(input("Enter the 1st num: "))
b = float(input("Enter the 2nd num: "))
op = input("Enter operator (+, -, *, /, %, **): ")

if op == '+':
    print(a + b)
elif op == '-':
    print(a - b)
elif op == '*':
    print(a * b)
elif op == '/':
    print(a / b)
elif op == '%':
    print(a % b)
elif op == '**':
    print(a ** b)
else:
    print("invalid")
```

```python
#students marks

sub1 = float(input("Enter marks of subject 1: "))
sub2 = float(input("Enter marks of subject 2: "))
sub3 = float(input("Enter marks of subject 3: "))

total = sub1 + sub2 + sub3
percentage = total / 3

print("Total marks:", total)
print("Percentage:", percentage)

if percentage >= 50:
    print("Result: Pass")
else:
    print("Result: Fail")
```

```python
#favorite fruits

fruits = ["🍏","🍊", "🍍", "🍑", "🍉"]

user = input("Enter a fruit: ")

if user in fruits:
    print(user, "exists in the list")
else:
    print(user, "does not exist in the list")
```
