```python
import pandas as pd

data = {
    "name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain","Irha", "Fatima", "Minha", "Nasir"],
    "course": ["AI", "Web", "AI", "Web", "AI", "Tech", "AI", "Web", "Tech"],
    "Marks": [45, 90, 75, 85, 40, 80, 55, 64, 99]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())

print("\nLowest Marks:")
print(df["Marks"].min())

print("________________________")
passed = df[df["Marks"] >= 50]
failed = df[df["Marks"] < 50]

print("\nAverage Marks:")
print(df[df["course"].isin(["Tech", "Web", "AI"])])

print("\nPassed Students:")
print(passed)

print("\nFailed Students:")
print(failed)

df["Result"] = df["Marks"].apply(
    lambda marks: "Pass" if marks >= 50 else "Fail"
)

print("-------------------------------")
print("\nFinal Data:")
print(df)

print("\nCourse Wise Average:")
print(df.groupby("course")["Marks"].mean())
```
