```python
import numpy as np
#TASK 01

marks = np.array([10, 20, 30, 40, 50, 60])

print("Sum:", np.sum(marks))
print("Average:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
```

```python
import numpy as np
#TASK 4
marks = np.random.randint(40, 101, 12)

print("Marks:", marks)
print("Average:", np.mean(marks))
print("Highest marks:", np.max(marks))
print("Lowest marks:", np.min(marks))
```

```python
import numpy as np
#TASK 5

marks = np.array([30, 67, 44, 45, 90, 25, 70, 95, 50, 55, 66, 99])

passed = marks[marks >= 50]
failed = marks[marks < 50]
topper = marks[marks > 90]

print("Passed marks:", passed)
print("Failed marks:", failed)
print("Topper marks:", topper)
print("-----------------------------")
print("Number of topper students:", len(topper))
print("Number of passed students:", len(passed))
print("Number of failed students:", len(failed))

```
