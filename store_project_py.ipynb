```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Store": ["Zullbery"] * 16,

    "Category": [
        "Fashion", "Fashion", "Fashion",
        "Fashion", "Fashion", "Fashion",
        "Accessories", "Accessories", "Accessories", "Accessories",
        "Bags", "Bags", "Bags",
        "Perfume", "Perfume",
        "Style"
    ],

    "Product": [
        "Dress", "Dress", "Dress",
        "Abaya", "Scarf", "Sunglasses",
        "Jewellery", "Jewellery", "Jewellery", "Jewellery",
        "Bag", "Bag", "Bag",
        "Perfume", "Perfume",
        "Watch"
    ],

    "Variant": [
        "Casual Dress", "Formal Dress", "Party Dress",
        "Classic Abaya", "Silk Scarf", "Classic Sunglasses",
        "Necklace", "Bracelet", "Earrings", "Ring",
        "Handbag", "Shoulder Bag", "Tote Bag",
        "Musk", "Oud",
        "Classic Watch"
    ],

    "Price": [
        4000, 5500, 7000,
        6500, 2500, 3500,
        2000, 2200, 1800, 2500,
        8000, 6000, 4500,
        2500, 4000,
        7500
    ],

    "Stock": [
        15, 8, 14,
        7, 15, 10,
        9, 10, 12, 10,
        6, 8, 15,
        15, 8,
        5
    ],

    "Remaining": [
        3, 3, 3,
        3, 4, 4,
        6, 7, 6, 4,
        2, 3, 3,
        3, 2,
        2
    ]
}

df = pd.DataFrame(data)

# Calculate how many units were sold
df["Sold"] = df["Stock"] - df["Remaining"]

print("ZULLBERY STORE DATA")
print(df)
```

```python
# Calculate sold units
df["Sold"] = df["Stock"] - df["Remaining"]

# Calculate revenue
df["Revenue"] = df["Price"] * df["Sold"]

# Display the detailed data
print(df[["Product", "Variant", "Price", "Sold", "Remaining", "Revenue"]])

# Display totals at the end
print("\n------------------------------------------")
print("TOTALS")
print("------------------------------------------")

print("Total Stock:", df["Stock"].sum())
print("Total Sold:", df["Sold"].sum())
print("Total Remaining:", df["Remaining"].sum())
print("Total Revenue: PKR", df["Revenue"].sum())
```

```python
# Best-selling products
best_selling_products = (
    df.groupby("Product")["Sold"]
      .sum()
      .sort_values(ascending=False)
)

print("BEST-SELLING PRODUCTS")
print("---------------------")
print(best_selling_products)
```

```python
# Revenue by variant
revenue_by_variant = (
    df.groupby("Variant")["Revenue"]
      .sum()
      .sort_values(ascending=False)
)

print("REVENUE BY VARIANT")
print("------------------")
print(revenue_by_variant)

print("===========================")
print()
print("ZULLBERY STOCK SUMMARY")
print("----------------------")

print("Total Stock:", df["Stock"].sum())
print("Total Sold:", df["Sold"].sum())
print("Total Remaining:", df["Remaining"].sum())
print("===========================")
print()

best_products = (
    df.groupby("Product")["Sold"]
      .sum()
      .sort_values(ascending=False)
)

print("BEST-SELLING PRODUCTS")
print("---------------------")
print(best_products)
```

```python
# Zullbery Business Analysis

print("ZULLBERY BUSINESS ANALYSIS")
print("==========================")

# 1. Best-selling products
best_products = df.groupby("Product")["Sold"].sum().sort_values(ascending=False)

print("\n1. UNITS SOLD BY PRODUCT")
print(best_products)

# 2. Revenue by product
product_revenue = df.groupby("Product")["Revenue"].sum().sort_values(ascending=False)

print("\n2. REVENUE BY PRODUCT")
print("---------------------")
print()
print(product_revenue)

# 3. Revenue by category
category_revenue = df.groupby("Category")["Revenue"].sum().sort_values(ascending=False)

print("\n3. REVENUE BY CATEGORY")
print("---------------------")
print()
print(category_revenue)

# 4. Total revenue
total_revenue = df["Revenue"].sum()

print("\n4. TOTAL REVENUE")
print("PKR", total_revenue)

# 5. Inventory value remaining
inventory_value = (df["Price"] * df["Remaining"]).sum()

print("\n5. REMAINING INVENTORY VALUE")
print("PKR", inventory_value)

# 6. Low-stock products
low_stock = df[df["Remaining"] <= 3][["Product", "Variant", "Remaining"]]

print("\n6. LOW-STOCK ITEMS")
print("-------------------------------------")
print(low_stock)
```

```python
import numpy as np

# Zullbery Sales Statistics

sold = df["Sold"].to_numpy()
revenue = df["Revenue"].to_numpy()

print("ZULLBERY SALES STATISTICS")
print("===========================")

print("Average units sold:", np.mean(sold))
print("Median units sold:", np.median(sold))
print("Highest units sold:", np.max(sold))
print("Lowest units sold:", np.min(sold))

print("\nAverage revenue per variant: PKR", np.mean(revenue))
print("Highest variant revenue: PKR", np.max(revenue))
print("Lowest variant revenue: PKR", np.min(revenue))
```

```python
feedback = input("Enter customer feedback (maximum 60 characters): ")

if len(feedback) > 60:

    print("Feedback is too long. Please keep it within 60 characters.")

else:

    feedback = feedback.lower()

    strong_found = []
    normal_found = []
    expression_found = []
    negative_word_found = []
    negative_expression_found = []

    for word in strong_positive:
        if word in feedback:
            strong_found.append(word)

    for word in normal_positive:
        if word in feedback:
            normal_found.append(word)

    for phrase in positive_expressions:
        if phrase in feedback:
            expression_found.append(phrase)

    for word in negative_words:
        if word in feedback:
            negative_word_found.append(word)

    for phrase in negative_expressions:
        if phrase in feedback:
            negative_expression_found.append(phrase)

    positive_found = strong_found + normal_found + expression_found
    negative_found = negative_word_found + negative_expression_found

    if positive_found and negative_found:
        result = "Mixed"
    elif positive_found:
        result = "Positive"
    elif negative_found:
        result = "Negative"
    else:
        result = "Neutral"

    print("\nZULLBERY CUSTOMER FEEDBACK")
    print("--------------------------")
    print("Comment:", feedback)
    print("Strong positive:", strong_found)
    print("Normal positive:", normal_found)
    print("Positive expressions:", expression_found)
    print("Negative words:", negative_word_found)
    print("Negative expressions:", negative_expression_found)
    print("OVERALL:", result)
```

```python
feedback_results = []

for comment in feedback_data:
    feedback = comment.lower()

    strong_found = []
    normal_found = []
    expression_found = []
    negative_word_found = []
    negative_expression_found = []

    for word in strong_positive:
        if word in feedback:
            strong_found.append(word)

    for word in normal_positive:
        if word in feedback:
            normal_found.append(word)

    for phrase in positive_expressions:
        if phrase in feedback:
            expression_found.append(phrase)

    for word in negative_words:
        if word in feedback:
            negative_word_found.append(word)

    for phrase in negative_expressions:
        if phrase in feedback:
            negative_expression_found.append(phrase)

    positive_found = strong_found + normal_found + expression_found
    negative_found = negative_word_found + negative_expression_found

    if positive_found and negative_found:
        result = "Mixed"
    elif positive_found:
        result = "Positive"
    elif negative_found:
        result = "Negative"
    else:
        result = "Neutral"

    feedback_results.append({
        "Comment": comment,
        "Positive": positive_found,
        "Negative": negative_found,
        "Overall": result
    })

print("ZULLBERY CUSTOMER FEEDBACK ANALYSIS")
print("------------------------------------")

for item in feedback_results:
    print("\nComment:", item["Comment"])
    print("Positive:", item["Positive"])
    print("Negative:", item["Negative"])
    print("Overall:", item["Overall"])
```

```python
print("CUSTOMER FEEDBACK SUMMARY")
print("=========================")

positive_count = 0
negative_count = 0
mixed_count = 0
neutral_count = 0

for item in feedback_results:
    if item["Overall"] == "Positive":
        positive_count += 1
    elif item["Overall"] == "Negative":
        negative_count += 1
    elif item["Overall"] == "Mixed":
        mixed_count += 1
    else:
        neutral_count += 1

print("Positive:", positive_count)
print("Negative:", negative_count)
print("Mixed:", mixed_count)
print("Neutral:", neutral_count)
```
